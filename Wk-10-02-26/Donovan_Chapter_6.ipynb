**Homework #1: Road Trip Playlist**

Scenario: You're making a playlist for a road trip and want to save it so it's still there tomorrow. Write **two void functions**:

* `write_playlist(filename, num_songs)` — opens the file in **write** mode and uses a `for` loop to ask the user for `num_songs` song titles, writing **each title on its own line**.
* `show_playlist(filename)` — opens the file in **read** mode and uses a `for` loop to print every song with a number in front of it. Use `rstrip('\n')` so there are **no blank lines** between songs.

In `main()`:
1. Ask the user how many songs they want, call `write_playlist`, then call `show_playlist`.
2. Ask the user for **one more** song and add it to the end of the file using **append** mode (don't erase the others!).
3. Call `show_playlist` again to show the updated playlist.

Example output (user input shown after the `?` and `:`):
```
How many songs? 3
Song 1: Life is a Highway
Song 2: Fast Car
Song 3: Take Me Home, Country Roads

Your playlist:
1. Life is a Highway
2. Fast Car
3. Take Me Home, Country Roads

Add one more song: Mr. Blue Sky

Your playlist:
1. Life is a Highway
2. Fast Car
3. Take Me Home, Country Roads
4. Mr. Blue Sky
```

````python
'''
Type Pseudocode here

'''
def write_playlist(filename, num_songs):
    with open(filename, "w") as file:
        for i in range(num_songs):
            song = input(f"song {i+1}: ")
            file.write(song + "\n")


def show_playlist(filename):
    with open(filename, "r") as file:
        print("\nYour playlist:")
        number = 1
        for line in file:
            song = line.rstrip("\n")
            print(f"{number}. {song}")
            number += 1


def main():
    filename = "playlist.txt"
    num_songs = int(input("How many songs do you want in your playlist? "))
    write_playlist(filename, num_songs)
    show_playlist(filename)
    new_song = input("\nAdd one more song: ")
    with open(filename, "a") as file:
        file.write(new_song + "\n")
    show_playlist(filename)


main()
````

Scenario: A weather station saves one daily high temperature per line in `temps.txt` (created by the setup cell). Write a value-returning function:

* `analyze_temps(filename)` — reads the file with a `for` loop and **returns four values**: the **number of days**, the **average** temperature, the **highest** temperature, and the **lowest** temperature. Don't forget to convert each line with `float()`.
  *Hint for highest/lowest:* use the **first** temperature as the starting value for both, then compare every temperature after it with an `if` statement.

And a void function:

* `write_report(report_filename, days, average, highest, lowest)` — **writes** the four results to a new file called `temp_report.txt`, one per line, with 1 decimal place.

In `main()`, call both functions, then open `temp_report.txt` for reading and **print its contents** so you can check your work.

Expected output:
```
Days recorded: 7
Average high: 73.4
Hottest day: 80.1
Coolest day: 68.0
```

````python
'''
analyze 
read 
calculate 4 things 
days, average, highest, lowerst
retrun those 
wrtie the report
receice values 
wrtie into the fil e
main()
call analyze
wrtie report
open it 
print it 

'''
def analyze_temps(filename):
    days = 0
    total = 0
    with open(filename, "r") as file:
        for line in file:
            temperature = float(line)
            if days == 0:
                highest = temperature 
                lowest = temperature 
            else:
                if temperature > highest:
                    highest = temperature
                if temperature < lowest:
                    lowest = temperature 
            total += temperature
            days += 1
    average = total / days 
    return days, average, highest, lowest 

def write_report(report_filename, days, average, highest, lowest):
    with open(report_filename, "w") as file:
        file.write(f"Days: {days}\n")
        file.write(f"Average Temperature: {average:.1f}\n")
        file.write(f"Highest Temperature: {highest:.1f}\n")
        file.write(f"Lowest Temperature: {lowest:.1f}\n")
def main():
    days, average, highest, lowest = analyze_temps("temps.txt")  
    write_report("report.txt", days, average, highest, lowest)
    with open("report.txt", "r") as file: 
        contents = file.read()
    print(contents)

main()
````

**Homework #3: Campus Store Inventory (Records)**

Scenario: The campus store keeps its inventory in `inventory.txt` (created by the setup cell). Each **record** has **three fields**, each on its own line: the **item name**, the **quantity** in stock (an integer), and the **price** (a float).

```

54.99            ← price
Ralphie Mug      ← record 2: item
...
```

Write two functions:

* `show_inventory(filename)` — reads **every record** using a `while` loop with the **empty-string sentinel** (`while item != ''`). For each record, print the item, the quantity, the price, and the **inventory value** (quantity × price) in lined-up columns. After the loop, print the **total inventory value** of the whole store.
* `find_item(filename, search_item)` — searches the file for a record whose item name **matches** `search_item`. If it finds one, **return** the quantity and price (**two values**). If it doesn't, **return `None`** — note that `return None` sends back a single value, so check the result with `if result is None:` *before* unpacking it.

In `main()`, call `show_inventory`, then ask the user for an item name and use `find_item` to print its quantity and price, or a "not found" message.

Example output:
```

Buff Hoodie         24  $  54.99  $  1,319.76
Ralphie Mug         40  $  14.50  $    580.00
Sticker Pack       150  $   4.25  $    637.50
```

```
Enter an item to look up: Foam Finger

```
💡 **Formatting hint:** `f'{item:<16}{qty:>6}'` left-aligns `item` in 16 spaces and right-aligns `qty` in 6.

````python
'''
show inventory, open file, set toal inventory print 
read first item 
while statement 
porint total
'''
def show_inventory(filename):
    total_value = 0
    with open(filename, "r") as file: 
        print(f"{"Item":<16}{"Qty":>6}{"Price":>10}{"Value":>12}")
        item = file.readline()
        while item != "":
            item = item.rstrip("\n")
            quantity = int(file.readline())
            price = float(file.readline())
            value = quantity * price 
            total_value += value 
            print(f"{item:<16}{quantity:>6}{price:>6.2f}{value:>9,.2f}")
            item = file.readline()
    print(f"Total inventory value: ${total_value:,.2f}")

def find_item(filename, search_item): 
    with open(filename, "r") as file: 
        item = file.readline()
        while item != "": 
            item = item.rstrip("\n") 
            quantity = int(file.readline())
            price = float(file.readline())
            if item == search_item: 
                return quantity, price
            item = file.readline()
        return None

def main():
    filename = "inventory.txt"
    show_inventory(filename)
    search_item = input("\nEnter an item to look up: ")
    result = find_item(filename, search_item)
    if result is not None:
        quantity, price = result 
        print(f"{search_item}: {quantity} in stock at ${price:.2f} each")
    else:
        print(f"Sorry, {search_item} is not in the inventory.")


main()
    
````

**Homework #4: Crash-Proof Sales Totaler**

Scenario: A manager has a folder full of weekly sales files, but some of them contain typos (like `N/A` instead of a number) and sometimes the manager mistypes a filename. Your program must **never crash**.

Write a function `total_sales_file(filename)` that uses **all four** parts of a `try` statement:
* **`try`:** open the file and use a `for` loop to add up the amounts and count the lines.
* **`except FileNotFoundError`:** print a message that the file could not be found.
* **`except ValueError as err`:** print a message that the file has bad data, **including the default error message** stored in `err`.
* **`else`:** runs only if nothing went wrong: print the number of days, the total (with commas and 2 decimals), and the average per day.
  *(Notice the `\n` in the ValueError message below: the bad line still had its newline when `float()` tried to convert it.)*
* **`finally`:** always print `Finished processing <filename>.`

In `main()`, use a `while` loop so the user can check **as many files as they want** (keep asking until they type `done`). Test it with `sales_week1.txt`, `sales_week2.txt`, and a filename that doesn't exist.

Example output:
```
Enter a sales file name (or 'done' to quit): sales_week1.txt
Days: 5
Total sales: $6,386.80
Average per day: $1,277.36
Finished processing sales_week1.txt.

Enter a sales file name (or 'done' to quit): sales_week2.txt
sales_week2.txt contains bad data: could not convert string to float: 'N/A\n'
Finished processing sales_week2.txt.

Enter a sales file name (or 'done' to quit): sales_week3.txt
Error: sales_week3.txt could not be found.
Finished processing sales_week3.txt.

Enter a sales file name (or 'done' to quit): done
Goodbye!
```

````python
'''
Type Pseudocode here

'''
def total_sales_file(filename):
    try:
        total = 0
        days = 0 
        with open(filename, "r") as file: 
            for line in file: 
                sales = float(line)
                total += sales 
                days += 1 
    except FileNotFoundError:
        print(f"Error: {filename} could not be found")
    except ValueError as err:
        print(f"{filename} contains invalid data: {err}")
    else:
        average = total / days 
        print(f"Days: {days}")
        print(f"Total sales: ${total:,.2f}")
        print(f"Average per day: ${average:,.2f}")
    finally:
        print(f"Finshed processing {filename}.")

def main():
    filename = input("Enter a sales file mame (or 'done' to quit): ")
    while filename != "done":
        total_sales_file(filename) 
        filename = input("\nEnter a sales files name (or 'done' to quit):")
    print("Goodbye!")

main()
````

**Homework #5 (Challenge): Personal Expense Tracker**

Scenario: You want to track your spending across the semester. Every expense is saved as a **record** with **two fields** in `expenses.txt`: a **description** and an **amount**. Because the data is in a file, your expenses are still there the next time you run the program!

Build the program from these functions:

* A global constant `FILENAME = 'expenses.txt'` at the top of the program.
* `show_menu()` — a **void** function that prints the menu:
  ```
  1. Add an expense
  2. View all expenses
  3. Show summary
  4. Quit
  ```
* `get_amount(prompt)` — asks the user the question in `prompt` and uses a `while` loop with **`try`/`except ValueError`** to **keep asking** until the user enters a valid number that is **greater than 0**. **Returns** the amount as a `float`.
* `add_expense(filename)` — asks for a description, calls `get_amount()`, and **appends** the record (two lines) to the file.
* `view_expenses(filename)` — reads and prints every record in lined-up columns. If the file doesn't exist yet, catch the **`FileNotFoundError`** and print `No expenses recorded yet.` instead of crashing.
* `summarize(filename)` — **returns three values**: the **number of expenses**, the **total spent**, and the **largest expense** amount. If the file doesn't exist, return `0, 0.0, 0.0`.
* `main()` — shows the menu in a `while` loop until the user chooses `4`. Print `Invalid choice.` for anything other than 1–4. For choice 3, print the summary (money with commas and 2 decimals, plus the **average** expense — watch out for dividing by zero when there are no expenses!).

Example run (starting with **no** `expenses.txt` file):

*(The `...` marks where the menu is printed again.)*

💡 **Testing tip:** To start over with an empty tracker, delete `expenses.txt` from the Files panel, or run `open('expenses.txt', 'w').close()` in a new cell.

````python
'''
Type Pseudocode here

'''
FILENAME = 'expenses.txt'


def show_menu():
    print('1. Add an expense')
    print('2. View all expenses')
    print('3. Show summary')
    print('4. Quit')


def get_amount(prompt):
    while True:
        try:
            amount = float(input(prompt))

            if amount > 0:
                return amount
            else:
                print('The amount must be greater than 0.')

        except ValueError:
            print('Please enter a valid number.')


def add_expense(filename):
    description = input('Description: ')
    amount = get_amount('Amount: $')

    with open(filename, 'a') as file:
        file.write(description + '\n')
        file.write(str(amount) + '\n')

    print(f'Saved: {description} ${amount:.2f}')


def view_expenses(filename):
    try:
        with open(filename, 'r') as file:
            description = file.readline()

            while description != '':
                description = description.rstrip('\n')
                amount = float(file.readline())

                print(f'{description:<20} $ {amount:>8.2f}')

                description = file.readline()

    except FileNotFoundError:
        print('No expenses recorded yet.')


def summarize(filename):
    count = 0
    total = 0.0
    largest = 0.0

    try:
        with open(filename, 'r') as file:
            description = file.readline()

            while description != '':
                amount = float(file.readline())

                count += 1
                total += amount

                if amount > largest:
                    largest = amount

                description = file.readline()

    except FileNotFoundError:
        return 0, 0.0, 0.0

    return count, total, largest


def main():
    choice = ''

    while choice != '4':
        show_menu()
        choice = input('Enter your choice: ')

        if choice == '1':
            add_expense(FILENAME)

        elif choice == '2':
            view_expenses(FILENAME)

        elif choice == '3':
            count, total, largest = summarize(FILENAME)

            if count > 0:
                average = total / count
            else:
                average = 0.0

            print(f'Expenses: {count}')
            print(f'Total spent: ${total:,.2f}')
            print(f'Average expense: ${average:,.2f}')
            print(f'Largest expense: ${largest:,.2f}')

        elif choice == '4':
            print('Goodbye!')

        else:
            print('Invalid choice.')

main()
````
